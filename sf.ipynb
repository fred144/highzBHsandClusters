### star formation


#### dealing with the GMC scales
recall the free-fall time $t_{\rm ff}$, defined as the minimum timescale for spherical object to undergo collapse under self gravity
$$t_{\rm ff} =  \left( \frac{3 \pi}{ 32 G \rho} \right)^{1/2}$$
the star formation efficiency (SFE) per free-fall time for a given gas cloud is 
$$\epsilon_{\mathrm{ff}}=\frac{\dot{M}_{*}}{M_{\mathrm{cloud}}/t_{\mathrm{ff}}}.$$
However, looking at the integrated SFE per cloud gives ([Chevance+22](https://arxiv.org/abs/2203.09570))
$$
\begin{aligned}\epsilon_{*}&\approx\quad\left(1+\frac{\Sigma_{\mathrm{crit}}}\Sigma\right)^{-1}\\\Sigma_{\mathrm{crit}}&\approx\quad\frac{\dot{p}/M_*}{\alpha_{\mathrm{vir}}G}=\alpha_{\mathrm{vir}}^{-1}\left(\frac{\dot{p}/M_*}{10\mathrm{~km~s}^{-1}\text{Муг}^{-1}}\right)\mathrm{~M_\odot~pc}^{-2}.\end{aligned}
$$


where 
- $\alpha_{\rm vir}$ is some virial parameters $\alpha_\mathrm{vir}\equiv\frac{5\sigma_v^2R}{GM}\propto\frac{\sigma_0^2}{\Sigma_\mathrm{mol}}$
- $\dot{p}/M_*$ is some feedback parameterization from an __internal__ process 


$M_*$ is cluster mass

For a star cluster of luminosity L, the rate (per unit mass) of momentum input from direct radiation is
$$\frac{\dot{p}_{r a d}}{M_{*}}=\frac{L}{c M_{*}}=19\mathrm{km
\: s}^{-1}\mathrm{Myr}^{-1}\frac{\Psi}{1800\mathrm{erg\,s}^{-1}\,\mathrm{g}^{-1}}$$


For wind-driven bubbles
$$\frac{\dot{p}_{w}}{M_{*}}=880\frac{\mathrm{km \: s^{-1}}}{\mathrm{Myr}}\left(\frac{t}{\mathrm{Myr}}\right)^{2/5}\left(\frac{M_{*}}{10^{3}\mathrm{M_{\odot}}}\right)^{-1/5} \times\;\left(\frac{n_{\mathrm{H}}}{10^{2}\mathrm{cm}^{-3}}\right)^{1/5}$$

Dominant SNe feedback (3-40 Myr from the formation of massive stars).
$p_{\mathrm{SN}}/M_{\star}\ \sim 1\,-\,3\,\times\,10^{3}\,{\mathrm{~km\:s}}^{-1}$
