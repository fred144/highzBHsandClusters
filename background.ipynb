recent observations by jwst are illucidating the difference between the high-redshift ($z \gtrsim 6$) and low-redshift universe, we aim to investigate a pathway to SMBH growth.

### Goal

- characterize BH growth timescales using different assumptions for accretion (may be super eddington) and duty cycle (for BH characterizes how often accretion occurs), under different assumptions for seeding mass
- charecterize gas depletion timescale for SF and value for different duty cycles (for SF, characterizes burstiness)
- are SF and SMBH growth are competing for gas in the first galaxies. or can high density, low metallcitiy gas both facilitate the formation of stars and the growth of BHs? if so, what conditions (halo masses, etc.)

### SMBHs at high-z

- Long before JWST, we have known about SMBH at high-z, but mostly restframe UV bright ($M_{\rm UV} \lesssim -22$) sources (e.g., [Kulkarni et al. 2019](10.1093/mnras/stz1493) ) using ground-based UV spectroscopy. 
- This is congruent with the picture that SMBH are responsible for fully quenching the high mass end of the SMHM, so UV bright quasars exist in gas poor (red and dead) environments. 
- $10^8 - 10^9 {\rm M_\odot}$ hundreds of BHs have been detected  at $z \gtrsim 6$
- There a new class of objects: little red dots observed from JWST [Matthee et al 2024](doi.org/10.3847/1538-4357/ad2345).
    - very red rest-frame continuum (3000 $\AA$ - 1 $\mu m$) broad line regions (high, temperature, high velcity dispersion regions 1200 - 3700 km/s)
    - compact, point-like sources, with some $\lesssim 200$ pc ([Akins et al. 2024](https://arxiv.org/abs/2410.00949))
- anomalies 
    - abundant (~$20$  per cent) of broad line slected AGNs, z ~ 5-6.
    - when accounting for dust attentuation, they are ~10-100 times higher than UV bright quasars ([Green et al.](doi.org/10.3847/1538-4357/ad1e5f))
    - They grow earlier than expected (e.g., [Pacucci et al. 2024](https://doi.org/10.3847/1538-4357/ad3044))

### Star Formation at High Redshift

- at high-z, SF is expected to depart from the canconical K-S relation: $\Sigma_{SFR}\propto(\Sigma_{gas})^{n\sim 1.4}$ due to low-Z ($< 0.001$ Zsun)and high surface densities ($\Sigma_{gas} > 10^3 \:{\rm cm^{-3}}$).
- may folow instead $\Sigma_{SFR}\propto 10^{-12} \kappa_s (\Sigma_{gas})^{n\sim 1.4}$, for some bursty parameter $\kappa_s$ ([Ferrera et al 2019](http://dx.doi.org/10.1093/mnras/stz2031)).
